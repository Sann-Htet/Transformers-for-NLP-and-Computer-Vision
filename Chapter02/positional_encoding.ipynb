```python
import math
import torch
from torch import cosine_similarity

d_model = 512
```

```python
def positional_encoding(pos, pe):
    """
    Compute the positional encoding for a given position and encoding dimension.

    Args:
        pos (int): The position for which to compute the encoding.
        pe (int): The dimension of the positional encoding.
    Returns:
        list: A list containing the positional encoding values for the given position.
    """
    for i in range(0, 512, 2):
        pe[0][i] = math.sin(pos / (10000 ** ((2 * i) / d_model)))
        pe[0][i + 1] = math.cos(pos / (10000 ** ((2 * i) / d_model)))
    return torch.tensor(pe)
```

```python
pos = 2
pe = [[0] * d_model]
print(positional_encoding(pos, pe))
```

```python
cosine_similarity(positional_encoding(2, pe), positional_encoding(10, pe))
```
